## Executive Summary

This notebook implements an end-to-end regression modeling pipeline on the Diamonds dataset. The workflow begins by loading the dataset directly from its source CSV, followed by data cleaning steps that address physically impossible zero-valued dimensions, outlier removal via the IQR rule, and ordinal encoding of the qualitative `cut`, `color`, and `clarity` attributes so that their natural value hierarchy is preserved in the numeric feature space.

The modeling phase evaluates multicollinearity among predictors using Variance Inflation Factors, and applies Principal Component Analysis to compress the highly correlated spatial dimensions (`x`, `y`, `z`) into a single `spatial_size_index`. Four regression specifications are compared — a baseline linear model, a log-linear model, a log-linear model using the PCA-derived size index (Model 3), and a refined log-linear model that additionally drops `carat` to fully resolve a residual collinearity issue discovered between `carat` and the size index (Model 4) — with model quality assessed via R-squared, adjusted R-squared, AIC, BIC, and the F-statistic.

The selected model is validated with residual diagnostics (mean-zero check, a Q-Q plot for normality, and a fitted-vs-residuals plot for homoscedasticity), and — critically — with genuine out-of-sample evaluation against the 20% test split that was set aside in Part 1. This out-of-sample check surfaces a real limitation invisible from in-sample metrics alone: the Step 3 IQR outlier filter truncates the training data to diamonds of roughly 2 carats or less, so predictions degrade sharply on the larger, rarer stones present in the test set. Section 3 of this notebook translates the fitted coefficients into business-relevant insights about how carat weight, quality grade, and physical size drive diamond price, and discusses this and other practical limitations of the model.

### Step 3: Outlier Mitigation via the IQR Rule

Extreme outliers add variance and artificially shift the mean and standard deviation of Ordinary Least Squares estimations. We clean our ratio-scaled variables using a strict Interquartile Range boundary.


```python
# Statistical distribution summary for numerical columns
print("=== Numerical Column Descriptive Statistics ===")
display(df[num_cols].describe())

# Visualization of distribution and ranges using Boxplots
plt.figure(figsize=(15, 8))
for i, col in enumerate(num_cols, 1):
    plt.subplot(2, 4, i)
    # sns.histplot(x=df[col], color='skyblue',kde=True)
    ax = sns.boxplot(y=df[col], color='skyblue')
    ax.set_ylim(bottom=0)
    plt.title(f'Distribution of {col}')

plt.tight_layout()
plt.show()
```

```python
# Visualizing distributions for Carat and Spatial Dimensions (x, y, z)
plt.figure(figsize=(16, 10))

plot_cols = ['carat', 'x', 'y', 'z']
colors = ['skyblue', 'teal', 'orange', 'crimson']

for i, (col, color) in enumerate(zip(plot_cols, colors), 1):
    plt.subplot(2, 2, i)
    sns.histplot(df[col], kde=True, color=color, bins=30)
    plt.title(f'Histogram of {col.capitalize()}', fontsize=14)
    plt.xlabel(f'{col} value')
    plt.ylabel('Frequency')

plt.tight_layout()
plt.show()
```

ok there are outliers for sure

```python
def clean_outliers_iqr(data, columns):
    processed_df = data.copy()
    for col in columns:
        Q1 = processed_df[col].quantile(0.25)
        Q3 = processed_df[col].quantile(0.75)
        IQR = Q3 - Q1
        lower_limit = Q1 - 1.5 * IQR
        upper_limit = Q3 + 1.5 * IQR
        # Retain observations exclusively within mathematically valid boundaries
        processed_df = processed_df[(processed_df[col] >= lower_limit) & (processed_df[col] <= upper_limit)]
    return processed_df

continuous_variables = ['carat', 'depth', 'table', 'x', 'y', 'z']
df_clean = clean_outliers_iqr(df, continuous_variables)
print(f"\nDataFrame capacity after outlier mitigation: {df_clean.shape[0]} rows (Filtered out noise)")
```

```python
# Comparative Descriptive Statistics: Pre- vs Post-Processing
print("=== Descriptive Statistics: Pre-Processed (Original) ===")
display(df[num_cols].describe())

print("\n=== Descriptive Statistics: Post-Processed (Cleaned) ===")
display(df_clean[num_cols].describe())

# Visualizing the impact of cleaning with focused scaling
plt.figure(figsize=(18, 14))

for i, col in enumerate(['carat', 'depth', 'table', 'x', 'y', 'z'], 1):
    ax = plt.subplot(2, 3, i)

    temp_original = pd.DataFrame({col: df[col], 'Status': 'Original'})
    temp_cleaned = pd.DataFrame({col: df_clean[col], 'Status': 'Cleaned'})
    comparison_df = pd.concat([temp_original, temp_cleaned])

    sns.boxplot(data=comparison_df, x='Status', y=col, palette='Set2', hue='Status', legend=False)

    stats_orig = {'mean': df[col].mean(), 'median': df[col].median()}
    stats_clean = {'mean': df_clean[col].mean(), 'median': df_clean[col].median()}

    # Plotting mean/median lines
    ax.axhline(y=stats_orig['mean'], xmin=0, xmax=0.5, color='red', linestyle='--', alpha=0.7)
    ax.axhline(y=stats_orig['median'], xmin=0, xmax=0.5, color='black', linestyle=':', alpha=0.7)
    ax.axhline(y=stats_clean['mean'], xmin=0.5, xmax=1, color='red', linestyle='--', alpha=0.7)
    ax.axhline(y=stats_clean['median'], xmin=0.5, xmax=1, color='black', linestyle=':', alpha=0.7)

    # Labels for context
    ax.text(-0.52, stats_orig['mean'], f"{stats_orig['mean']:.2f}", color='red', ha='right', va='center', fontweight='bold', fontsize=9)
    ax.text(-0.52, stats_orig['median'], f"{stats_orig['median']:.2f}", color='black', ha='right', va='center', fontweight='bold', fontsize=9)
    ax.text(0.5, stats_clean['mean'], f"{stats_clean['mean']:.2f}", color='red', ha='center', va='bottom', fontsize=9)
    ax.text(0.5, stats_clean['median'], f"{stats_clean['median']:.2f}", color='black', ha='center', va='top', fontsize=9)

    # Scaling the charts to show only IQR and Whiskers
    # Based on the Cleaned data as it defines the 'valid' range
    q1 = df_clean[col].quantile(0.25)
    q3 = df_clean[col].quantile(0.75)
    iqr = q3 - q1

    # Setting Y-limit to precisely the theoretical whisker range (1.5 * IQR)
    ax.set_ylim(max(df_clean[col].min(), q1 - 2.55 * iqr), min(df_clean[col].max(), q3 + 2.55 * iqr))

    plt.title(f'Focused Distribution: {col.capitalize()}')
    plt.xlabel('')
    if i == 1:
        plt.plot([], [], color='red', linestyle='--', label='Mean')
        plt.plot([], [], color='black', linestyle=':', label='Median')
        plt.legend(loc='upper right')

plt.tight_layout()
plt.show()
```

As the comparison above shows, dropping rows outside 1.5×IQR bounds noticeably shifts the mean, median, and standard deviation of `carat`, `x`, `y`, and `z` relative to the original data — most visibly, it caps the training distribution at roughly **2 carats**. We accept this trade-off deliberately: OLS is highly sensitive to extreme-leverage points, and these observations dominate the boxplots above by a wide margin. We proceed with the IQR-filtered frame (`df_clean`) as the modeling dataset for the rest of this notebook.

This decision is not free, however — it means the model never sees diamonds above ~2 carats during training. We flag this explicitly now and return to quantify its real-world consequence with genuine held-out test data in **Step 4b** of Part 2, where it turns out to matter a great deal.

The IQR filter dropped ~3,675 rows (roughly 4k) and materially changed the dataset's summary statistics, as seen above. Before settling on `df_clean` as the modeling dataset, we explore an alternative that doesn't discard any rows: `RobustScaler`, which rescales using the median and IQR instead of the mean and standard deviation, making it inherently less sensitive to outliers than z-score standardization.

```python
from sklearn.preprocessing import RobustScaler

# Initialize the RobustScaler
robust_scaler = RobustScaler()

# Select the continuous numerical columns to scale
scale_cols = ['carat', 'depth', 'table', 'x', 'y', 'z']

# Create a new dataframe for robust scaled data
df_robust = df.copy()
df_robust[scale_cols] = robust_scaler.fit_transform(df[scale_cols])

print("=== Robust Scaling Complete ===")
print(f"New Dataframe Shape: {df_robust.shape}")
display(df_robust[scale_cols].describe())
display(df_robust.head())
```

```python
# Identify numerical columns used in scaling
numeric_cols = ['carat', 'depth', 'table', 'x', 'y', 'z']
```

```python


# 1. Create a new dataframe where outliers are clipped to the IQR boundaries
df_robust_clipped = df_robust.copy()

for col in numeric_cols:
    Q1 = df_robust_clipped[col].quantile(0.25)
    Q3 = df_robust_clipped[col].quantile(0.75)
    IQR = Q3 - Q1
    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    # Clipping values outside the [lower_bound, upper_bound] range
    df_robust_clipped[col] = df_robust_clipped[col].clip(lower=lower_bound, upper=upper_bound)

print(f"Clipped dataframe created. Shape: {df_robust_clipped.shape}")
display(df_robust_clipped[numeric_cols].describe())
```

**Decision:** `df_robust` and `df_robust_clipped` above are kept for comparison, but the final modeling pipeline does **not** use them. Two reasons: (1) `RobustScaler` rescales every leverage point rather than removing it, so the same extreme observations that distorted the boxplots in Step 3 would still directly influence the OLS fit; and (2) working in the scaled units would make the regression coefficients harder to interpret in Part 3 (a one-unit change would no longer correspond to one carat or one millimeter). The rest of this notebook — VIF, PCA, and all regression models — uses `df_clean`, the row-filtered frame from Step 3.

### Step 4: True Ordinal Mapping

Attributes such as `cut`, `color`, and `clarity` are ordered qualitative steps, not arbitrary nominal labels. We map them to numeric vectors that mirror their real-world value hierarchy.


```python
ordinal_mappings = {
    'cut': {'Fair': 0, 'Good': 1, 'Very Good': 2, 'Premium': 3, 'Ideal': 4},
    'color': {'J': 0, 'I': 1, 'H': 2, 'G': 3, 'F': 4, 'E': 5, 'D': 6},
    'clarity': {'I1': 0, 'SI2': 1, 'SI1': 2, 'VS2': 3, 'VS1': 4, 'VVS2': 5, 'VVS1': 6, 'IF': 7}
}

for col, mapping in ordinal_mappings.items():
    df_clean[col] = df_clean[col].map(mapping)

# Confirm no unmapped null elements remain
assert df_clean[['cut', 'color', 'clarity']].isnull().sum().sum() == 0
```

## Part 2: Model Development and Validation (10 Marks)

### Step 1: Inspecting Collinearity and Calculating VIF

We calculate Variance Inflation Factors to quantify how much the variance of our estimated regression coefficients is inflated due to collinearity.


```python
X_vif_check = df_clean[['carat', 'cut', 'color', 'clarity', 'depth', 'table', 'x', 'y', 'z']]
X_vif_check = sm.add_constant(X_vif_check)

vif_summary = pd.DataFrame({
    "Predictor Feature": X_vif_check.columns,
    "Calculated VIF value": [variance_inflation_factor(X_vif_check.values, i) for i in range(X_vif_check.shape[1])]
})
print("=== Variance Inflation Factor (VIF) Output ===")
print(vif_summary.to_string(index=False))
```

### Step 2: Applying Principal Component Analysis (PCA)

Because a diamond's spatial metrics (`x`, `y`, `z`) correlate heavily with its overall mass (`carat`), they generate extreme multicollinearity. We use PCA to collapse these dimensions into a single uncorrelated "Size Index".


```python
spatial_features = ['x', 'y', 'z']
scaler = StandardScaler()
scaled_spatial_matrix = scaler.fit_transform(df_clean[spatial_features])

# Instantiate and construct a single overarching dimension
pca_engine = PCA(n_components=1, random_state=42)
df_clean['spatial_size_index'] = pca_engine.fit_transform(scaled_spatial_matrix)

print("\n=== Dimensionality Reduction Metrics ===")
print(f"Total variance explained by primary Spatial Component: {pca_engine.explained_variance_ratio_[0]:.2%}")
```

### Step 3: Multi-Model Regression Iterations

We build three models to systematically evaluate how structural changes and feature transformations alter our predictions.


```python
independent_base = ['carat', 'cut', 'color', 'clarity', 'depth', 'table']

# Model 1: Baseline Linear Model (Uses all features, exposed to collinearity distortions)
X_model1 = sm.add_constant(df_clean[independent_base + ['x', 'y', 'z']])
model1_fit = sm.OLS(df_clean['price'], X_model1).fit()

# Model 2: Log-Linear Model (Uses a log-transformed target to handle the heavily skewed price distribution)
y_log_transformed = np.log(df_clean['price'])
model2_fit = sm.OLS(y_log_transformed, X_model1).fit()

# Model 3: PCA-Optimized Model (Log-transformed target, with collinear x, y, z metrics replaced by our single size index)
X_model3 = sm.add_constant(df_clean[independent_base + ['spatial_size_index']])
model3_fit = sm.OLS(y_log_transformed, X_model3).fit()

print("\n=== Selected Model Summary (Model 3) ===")
print(model3_fit.summary())
```

### Step 3b: Verifying Multicollinearity Resolution Post-PCA

Model 3's summary above has a red flag: the `carat` coefficient is **negative** (-0.9685), which contradicts basic domain knowledge — bigger diamonds cost more, not less. A negative sign on an obviously important, positively-related predictor is a classic symptom of unresolved multicollinearity. PCA collapsed `x`, `y`, `z` into `spatial_size_index`, but it never looked at `carat` — and `carat` is essentially a measurement of the same physical property (size/mass) that `x`, `y`, `z` measure. We check whether that redundancy survived PCA by recomputing VIF on Model 3's actual design matrix.

```python
# Recompute VIF on Model 3's actual predictor set (post-PCA)
vif_post_pca = pd.DataFrame({
    "Predictor Feature": X_model3.columns,
    "Calculated VIF value": [variance_inflation_factor(X_model3.values, i) for i in range(X_model3.shape[1])]
})
print("=== VIF After PCA (Model 3 Predictors) ===")
print(vif_post_pca.to_string(index=False))

print(f"\nCorrelation(carat, spatial_size_index) = {df_clean['carat'].corr(df_clean['spatial_size_index']):.4f}")
```

**Finding:** `carat` and `spatial_size_index` both carry a VIF of **~37.9**, essentially unchanged from `carat`'s pre-PCA VIF of 37.97, and the two are correlated at **r = 0.986**. PCA successfully eliminated the collinearity *among* `x`, `y`, `z` (their individual VIFs dropped from the hundreds to a single clean component), but it did nothing for the deeper redundancy between `carat` and physical size — because carat and size are, physically, nearly the same quantity (mass ≈ density × volume). With two near-identical proxies for "how big is this diamond" both in the model, OLS can't tell them apart and starts assigning arbitrary, unstable, offsetting signs to their coefficients — which is exactly what produced `carat`'s implausible negative coefficient. **Model 4** below resolves this properly by keeping only one proxy for size.

```python
# Model 4: Log-Linear model that drops the redundant 'carat' term, keeping
# 'spatial_size_index' as the sole size proxy, to fully resolve the collinearity found above.
independent_no_carat = ['cut', 'color', 'clarity', 'depth', 'table', 'spatial_size_index']
X_model4 = sm.add_constant(df_clean[independent_no_carat])
model4_fit = sm.OLS(y_log_transformed, X_model4).fit()

vif_model4 = pd.DataFrame({
    "Predictor Feature": X_model4.columns,
    "Calculated VIF value": [variance_inflation_factor(X_model4.values, i) for i in range(X_model4.shape[1])]
})

print("=== Model 4 Summary (Carat Dropped) ===")
print(model4_fit.summary())
print("\n=== VIF Check: Model 4 ===")
print(vif_model4.to_string(index=False))
```

**Result:** Every VIF in Model 4 drops to roughly 1.0–1.5 — multicollinearity is fully resolved — and `spatial_size_index`'s coefficient is now a sensible **positive 0.598**, with every other coefficient sign matching domain expectations. The cost is a small drop in fit: R² falls from 0.9791 (Model 3) to 0.9749 (Model 4). We carry both Model 3 and Model 4 forward into the comparison matrix and, importantly, into out-of-sample testing below, since in-sample R² alone cannot tell us which one actually generalizes better.

### Step 4: Model Validation Evaluation Matrix

We generate a comparative statistical matrix to isolate and justify the best-performing model structure.


```python
validation_matrix = {
    "Model Specifications": ["Model 1: Baseline Linear", "Model 2: Log-Linear (All Fields)", "Model 3: Log-Linear + PCA Size Index", "Model 4: Log-Linear + PCA Size Index (Carat Dropped)"],
    "R-squared": [model1_fit.rsquared, model2_fit.rsquared, model3_fit.rsquared, model4_fit.rsquared],
    "Adjusted R-squared": [model1_fit.rsquared_adj, model2_fit.rsquared_adj, model3_fit.rsquared_adj, model4_fit.rsquared_adj],
    "AIC Metric": [model1_fit.aic, model2_fit.aic, model3_fit.aic, model4_fit.aic],
    "BIC Metric": [model1_fit.bic, model2_fit.bic, model3_fit.bic, model4_fit.bic],
    "F-Statistic p-value": [model1_fit.f_pvalue, model2_fit.f_pvalue, model3_fit.f_pvalue, model4_fit.f_pvalue]
}

df_validation = pd.DataFrame(validation_matrix)
print("\n=== Systematic Model Comparison Matrix ===")
print(df_validation.to_string(index=False))
```

**Reading this table correctly:** Model 2 has the best raw R²/AIC/BIC of all four, but we don't select it — it still carries the raw `x`, `y`, `z` fields, whose VIFs run into the hundreds (Step 1), meaning its individual coefficients are statistically unreliable even though its aggregate fit looks best. Between the two multicollinearity-aware candidates, Model 3 (R²=0.9791) edges out Model 4 (R²=0.9749) in-sample, but Model 3 still has the unresolved `carat`/`spatial_size_index` redundancy from Step 3b. In-sample metrics can't settle this trade-off on their own — **Step 4b** below tests both against the held-out test set to see which one actually generalizes.

### Step 4b: Out-of-Sample Validation on the Held-Out Test Set

Part 1 set aside 20% of the data (`test_df`) but nothing in the notebook so far has actually used it — every R², AIC, BIC, and F-statistic above is an **in-sample** fit statistic, which cannot detect overfitting or poor generalization. We now apply the *exact same fitted transformations* learned from the training data — the carat-bin median imputation lookup (`impute_lookup`, `carat_bin_edges`), the ordinal mappings, the `StandardScaler`, and the PCA projection — to `test_df`, and score Model 3 and Model 4 on genuinely unseen data. Critically, we do **not** apply the Step 3 IQR row-filter to the test set: dropping "hard" rows from the evaluation set would hide exactly the generalization failures we're trying to detect.

```python
from sklearn.metrics import r2_score, mean_squared_error, mean_absolute_error

# 1. Apply the same preprocessing pipeline fitted on the training data to test_df
test_processed = test_df.copy()
test_processed[dim_cols] = test_processed[dim_cols].replace(0, np.nan)

# Bin test carats using the TRAINING bin edges (no fitting on test data)
test_processed['carat_bin'] = pd.cut(test_processed['carat'], bins=carat_bin_edges)
for col in dim_cols:
    test_processed[col] = test_processed[col].fillna(test_processed['carat_bin'].map(impute_lookup[col]))
    # Fallback for test carats outside the training bin range
    test_processed[col] = test_processed[col].fillna(df[col].median())
test_processed.drop(columns=['carat_bin'], inplace=True)

# Apply the same ordinal mapping
for col, mapping in ordinal_mappings.items():
    test_processed[col] = test_processed[col].map(mapping)

# Apply the already-FITTED scaler and PCA (transform only, never re-fit on test data)
scaled_spatial_test = scaler.transform(test_processed[spatial_features])
test_processed['spatial_size_index'] = pca_engine.transform(scaled_spatial_test)

# 2. Score Model 3 and Model 4 out-of-sample
results = []
for name, model, features in [("Model 3", model3_fit, independent_base + ['spatial_size_index']),
                               ("Model 4", model4_fit, independent_no_carat)]:
    X_test = sm.add_constant(test_processed[features], has_constant='add')[model.model.exog_names]
    pred_log = model.predict(X_test)
    actual_log = np.log(test_processed['price'])
    pred_price = np.exp(pred_log)
    actual_price = test_processed['price']

    results.append({
        "Model": name,
        "Train R2 (log)": model.rsquared,
        "Test R2 (log)": r2_score(actual_log, pred_log),
        "Test RMSE (log)": np.sqrt(mean_squared_error(actual_log, pred_log)),
        "Test R2 (price $)": r2_score(actual_price, pred_price),
        "Test RMSE (price $)": np.sqrt(mean_squared_error(actual_price, pred_price)),
        "Test MAE (price $)": mean_absolute_error(actual_price, pred_price),
    })

df_test_results = pd.DataFrame(results)
print("=== Out-of-Sample Test Set Performance ===")
print(df_test_results.to_string(index=False))

# 3. Isolate the large-carat rows the training set (capped at ~2 carats by the IQR filter) never saw
large_carat_test = test_processed[test_processed['carat'] > df_clean['carat'].max()]
print(f"\nTest-set rows with carat > {df_clean['carat'].max():.2f} (training max): {len(large_carat_test)} of {len(test_processed)}")
```

**This is the most important finding in the notebook.** On the log-price scale, both models generalize fine (Test R² ≈ 0.978 for Model 3, ≈ 0.965 for Model 4 — close to their training R²). But converted back to actual dollars, the two models diverge sharply:

| Model | Test R² (price $) | Test RMSE (price $) | Test MAE (price $) |
|---|---|---|---|
| Model 3 (carat + size index) | ≈ 0.94 | ≈ $976 | ≈ $467 |
| Model 4 (size index only) | ≈ 0.45 | ≈ $2,960 | ≈ $835 |

Model 4 — the model we just showed to be the statistically "correct" one, with clean VIFs and sensible coefficient signs — collapses on the price scale, driven almost entirely by the **379 test-set diamonds above ~2 carats** printed above. Since `df_clean` (Step 3's IQR filter) excluded every training diamond above 2 carats, neither model ever learned that part of the price curve. Model 3 happens to look more robust here, but for the wrong reason: its two collinear, opposite-signed coefficients on `carat` (-0.97) and `spatial_size_index` (+0.82) partially cancel each other out when extrapolating beyond the training range, which accidentally dampens (rather than genuinely fixes) the blow-up. Model 4 has no such offsetting term, so its single `spatial_size_index` coefficient extrapolates unchecked and produces predictions up to $100k+ for diamonds that actually sell for ~$15k.

**Root cause:** this traces directly back to the Step 3 decision to IQR-filter the training data. Those large-carat rows were not measurement errors — they were legitimate, valid transactions removed for being statistically "extreme." The IQR rule is not a good fit for a variable like `carat` whose real-world distribution is genuinely long-tailed. We revisit this in Part 3's limitations and recommendations.

### Step 5: Running Regression Diagnostics

Given Step 4b's finding, we proceed with **Model 3** as the primary model for prediction accuracy within the observed data range (it substantially outperforms Model 4 on out-of-sample dollar error), while treating **Model 4** as the reference model for interpreting individual coefficients cleanly. We plot and evaluate Model 3's residuals ($\epsilon = y - \hat{y}$) to verify that it satisfies standard OLS assumptions.

```python
optimal_residuals = model3_fit.resid

print("\n=== Residual Diagnostic Analytics (Model 3) ===")
print(f"Residual Mean Value: {np.mean(optimal_residuals):.6f} (Should converge to zero)")
print(f"Residual Mean Value (Model 4, for reference): {np.mean(model4_fit.resid):.6f}")

# Create dual verification charts
fig, axes = plt.subplots(1, 2, figsize=(16, 5))

# Quantile-Quantile (Q-Q) Error Normality Chart
stats.probplot(optimal_residuals, dist="norm", plot=axes[0])
axes[0].set_title("Residual Normal Probability Plot (Q-Q Chart)")

# Homoscedasticity Variance Distribution Chart
sns.scatterplot(x=model3_fit.fittedvalues, y=optimal_residuals, alpha=0.15, ax=axes[1])
axes[1].axhline(0, color='crimson', linestyle='--', linewidth=2)
axes[1].set(title="Residual Errors vs. Fitted Estimates", xlabel="Predicted Values (Log Scale)", ylabel="Residual Error")

plt.tight_layout()
plt.savefig("part2_model_diagnostics.png", dpi=300)
plt.show()
```

## Part 3: Interpretation, Insights, and Structural Boundaries (4 Marks)

### 1. Translation of Beta Parameters ($\beta_j$)

Both Model 3 and Model 4 use a natural-log-transformed target ($\ln(\text{Price})$), so each coefficient $\beta_j$ translates to an approximate percentage price change via $(e^{\beta_j}-1)\times 100\%$ for a one-unit increase in that predictor, holding everything else constant.

- **Carat and Size are entangled in Model 3.** Model 3's `carat` coefficient (-0.9685 → an apparent **-62%** per additional carat) is **not causally meaningful on its own** — Step 3b showed `carat` and `spatial_size_index` share a VIF of ~38 and a correlation of 0.986, so OLS cannot separate their individual effects; it only assigned them offsetting signs because the two are jointly, not individually, identified. The combined effect of "diamond gets 1 unit bigger by both measures at once" is what matters, not either coefficient in isolation.
- **Model 4 gives the trustworthy size effect.** With `carat` removed and multicollinearity fully resolved (all VIFs ≈ 1.0–1.5), `spatial_size_index`'s coefficient of **0.5984** (≈ **+82%** price per one-standard-deviation increase in physical size) is the more reliable estimate of how strongly size drives price, since it isn't fighting a redundant predictor for credit.
- **Quality metrics are consistent and reliable across both models.** Moving up one ordinal grade is associated with: `clarity` **+12–13%**, `color` **+8–9%**, `cut` **+2–2.5%** (Model 3: 12.61%, 8.19%, 2.49%; Model 4: 12.12%, 8.58%, 2.43% respectively). These estimates barely move between the two specifications — evidence that, unlike `carat`, the quality grades are not entangled in the collinearity problem, and their marginal effect on price is genuinely estimated. Clarity has the largest quality-driven premium, followed by color, then cut.
- **Depth and table are minor factors.** Each contributes roughly **1–2%** per unit — statistically significant given the large sample size, but practically small next to size and clarity.

### 2. Regression Assumptions and Diagnostic Validation (Model 3)

- **Mean of Residuals:** 0.000000 — residuals are exactly zero-centered, confirming unbiased predictions on average (Model 4's residual mean is likewise ≈0, as expected for any OLS fit with an intercept).
- **Homoscedasticity:** The Fitted-vs-Residuals plot shows a roughly even error band across the prediction range with no strong funnel shape, broadly supporting constant-variance, though there is some visible fanning at the low end of fitted values.
- **Normality of Errors:** The Q-Q plot tracks the reference line closely through the middle of the distribution but deviates in the tails. This is confirmed numerically: Jarque-Bera test rejects normality (p ≈ 0.00), with a kurtosis of 7.41 (vs. 3.0 for a true normal) — the residuals are more heavy-tailed than OLS assumes. Durbin-Watson is 1.995, essentially ruling out first-order autocorrelation.
- **Multicollinearity:** Model 3's condition number (7.27e+03) and its own printed statsmodels warning confirm what Step 3b diagnosed directly — `carat` and `spatial_size_index` remain jointly, not individually, interpretable in this specification.

### 3. Model Boundaries and Analytical Limitations

- **The dominant limitation, proven with held-out data:** Step 4b showed that Step 3's IQR filter caps the training data at ~2 carats, and neither model ever learns the price behavior of larger stones. On the 379 test-set diamonds above that threshold, Model 3's price-scale R² and RMSE (≈0.94, ≈$976 overall) are respectable in aggregate only because its collinear coefficients coincidentally dampen extrapolation — Model 4, despite being the statistically cleaner model, degrades to R²≈0.45 and RMSE≈$2,960 on the same held-out set once large stones are included, because it extrapolates a single unchecked linear term past the range it was trained on. **Neither model should be trusted for diamonds larger than ~2 carats.**
- **Carat's individual effect cannot be isolated from size in Model 3.** Any headline like "a diamond loses 62% of its value per extra carat" (the raw Model 3 coefficient) would be a statistically indefensible claim given the multicollinearity documented in Step 3b — it must be read jointly with `spatial_size_index`, not alone.
- **Heavy-tailed residuals** (kurtosis 7.41) mean the model underestimates the frequency of large prediction errors relative to a normal-error assumption — consistent with the extrapolation failure above, but also present to a lesser degree within the training range.
- **The model is purely physical/quality-based** — it has no visibility into brand, certifying body, market timing, or buyer-specific demand, all of which are known to move diamond prices independent of the stone's measurable characteristics.

### 4. Strategic Recommendations

1. **Don't blanket-drop large-carat rows as "outliers."** Step 4b proved this was the single biggest driver of poor generalization. A better approach would be to keep all rows and either (a) fit a separate model or spline term for stones above some threshold, or (b) use a modeling approach that doesn't require symmetric-tail assumptions (e.g. quantile regression, or a tree-based model — see #3 below) rather than deleting genuine high-value transactions via a rule (1.5×IQR) that was designed for roughly-symmetric distributions, not a long-tailed one like carat weight.
2. **Pick one size proxy, not two.** If the goal is prediction accuracy within the observed range, Model 3's combined carat+size representation works empirically; if the goal is a defensible, individually-interpretable coefficient story (e.g., for a pricing policy document), use Model 4 and drop the redundant `carat` term entirely — don't report Model 3's individual `carat` coefficient as a standalone finding.
3. **Explore non-linear / tree-based models** (gradient boosting, random forests) that can flexibly capture the more step-like, rarity-driven pricing behavior at the high end of the carat range without requiring an explicit outlier-exclusion trade-off.
4. **Incorporate certifying-body and macro/time features** (GIA/AGS grading body, inflation-adjusted pricing indices) to explain price variance unrelated to the physical characteristics captured here.
